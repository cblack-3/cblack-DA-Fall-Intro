## If, Elif, Else

Write a function that generates a random number between 1 and 6 (inclusive). If the random number is 1, return the string 'One'. If the random number is 2, return the string 'Two'. For all other random numbers, return the string 'Neither one nor two'.

```python
def random_example():
    '''Generates a random number between 1 and 6 (inclusive). If the random number is 1, return the string...the question'''
    ran_num = np.random.choice(np.arange(1,6))

    if ran_num == 1:
        number = 'One'
    elif ran_num == 2:
        number = 'Two'
    else:
        number = 'Neither one nor two.'

    return number

random_example()
```

## For loops and append

Suppose a bag contains 3 red balls and 7 green balls. Use a for loop and sample_proportions to perform the following experiment 1000 times:
- draw a ball from the bag 10 times
- calculate the number of red balls drawn
- append the number of red balls to an array

After the loop completes, visualize the outcomes using a histogram.


```python
proportions = make_array(3/10, 7/10)
outcomes = make_array()

for _ in np.arange(1000):
    sample = sample_proportions(10, proportions)
    red = sample.item(0) * 10
    outcomes = np.append(outcomes, red)

results_tbl = Table().with_column('outcomes', outcomes)
results_tbl.hist()
```

## Solution

```python
bag_proportions = make_array(3/10, 7/10)
outcomes = make_array()

for _ in np.arange(1000):
    sample = sample_proportions(10, bag_proportions)
    num_red = sample.item(0) * 10
    outcomes = np.append(outcomes, num_red)

red_tbl = Table().with_column('Num Red', outcomes)
red_tbl.hist()
```
